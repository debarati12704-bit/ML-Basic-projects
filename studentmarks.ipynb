```python
import pandas as pd
```

```python
df=pd.read_csv('studentmarks.csv')
df.head()
```

✔ Find topper
✔ Find weakest subject
✔ Compare students
✔ Plot graphs
✔ Check attendance vs marks

```python
#find class average
total_marks=df["Math"].sum()+df["Science"].sum()+df["English"].sum()+df["Computer"].sum()
num_students=df.shape[0]
class_average=total_marks/num_students
print("Class Average Marks:", class_average)
```

```python
#adding total marks column
total_marks_per_student=df['Math']+df['Science']+df['English']+df['Computer']
df['Total Marks']=total_marks_per_student
df.head()
```

```python
#class topper
topper=df[df['Total Marks']==df['Total Marks'].max()]
print("Class Topper:", topper['Name'].iloc[0], "; with total marks:", topper['Total Marks'].iloc[0])
```

```python
#find weakest subject
avg_math=df['Math'].mean()
avg_science=df['Science'].mean()
avg_english=df['English'].mean()
avg_computer=df['Computer'].mean()

subject_avg = {
    'Math': avg_math,
    'Science': avg_science,
    'English': avg_english,
    'Computer': avg_computer
}


weakest_subject = min(subject_avg, key=subject_avg.get)

print("Weakest Subject:", weakest_subject)
print("Weakest Subject Average Marks:", subject_avg[weakest_subject])
```

```python
#comparing marks of two students
student1 = df[df['Name'] == 'Amit']    
student2 = df[df['Name'] == 'Arjun']
print("Comparing marks of Amit and Arjun:")
for subject in ['Math', 'Science', 'English', 'Computer']:
    if student1[subject].iloc[0] > student2[subject].iloc[0]:
        print(f"Amit scored higher than Arjun in {subject}.")
    elif student1[subject].iloc[0] < student2[subject].iloc[0]:
        print(f"Arjun scored higher than Amit in {subject}.")
    else:
        print(f"Amit and Arjun scored the same in {subject}.")
```

```python
#plot graphs
import matplotlib.pyplot as plt
subjects = ['Math', 'Science', 'English', 'Computer']
averages = [avg_math, avg_science, avg_english, avg_computer]  

plt.bar(subjects, averages, color=['blue', 'orange', 'green', 'red'])
plt.xlabel('Subjects')  
plt.ylabel('Average Marks')
plt.title('Average Marks per Subject')  
plt.legend(['Average Marks'])
plt.show()
```

```python
#Check attendance vs marks
plt.scatter(df['Attendance'], df['Total Marks'],color='purple')
plt.xlabel("Attendance (%)")
plt.ylabel("Total Marks")
plt.title("Attendance vs Total Marks")
plt.show()
```

```python
#find highest and lowest scorers:

math_topper = df[df['Math'] == df['Math'].max()]['Name'].iloc[0]
math_lowest = df[df['Math'] == df['Math'].min()]['Name'].iloc[0]
science_topper = df[df['Science'] == df['Science'].max()]['Name'].iloc[0]
science_lowest = df[df['Science'] == df['Science'].min()]['Name'].iloc[0]
english_topper = df[df['English'] == df['English'].max()]['Name'].iloc[0]
english_lowest = df[df['English'] == df['English'].min()]['Name'].iloc[0]
computer_topper = df[df['Computer'] == df['Computer'].max()]['Name'].iloc[0]
computer_lowest = df[df['Computer'] == df['Computer'].min()]['Name'].iloc[0]
print(f"Math Topper: {math_topper}, Math Lowest Scorer: {math_lowest}")
print(f"Science Topper: {science_topper}, Science Lowest Scorer: {science_lowest}")
print(f"English Topper: {english_topper}, English Lowest Scorer: {english_lowest}")     
print(f"Computer Topper: {computer_topper}, Computer Lowest Scorer: {computer_lowest}")

print(f"class topper: {topper['Name'].iloc[0]}, with total marks: {topper['Total Marks'].iloc[0]}")
lowest_score=df[df['Total Marks']==df['Total Marks'].min()]
print(f"lowest scorer: {lowest_score['Name'].iloc[0]}, with total marks: {lowest_score['Total Marks'].iloc[0]}")
```
